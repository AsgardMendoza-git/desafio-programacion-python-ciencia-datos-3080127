# Objetivo: reportar datos faltantes en un DataFrame de pandas

```python
import pandas as pd
```

```python
df = pd.read_csv('empleados.csv', delimiter=";")
```

```python
df.head()
```

```python
df.info()
```

Se puede observar que el DataFrame contiene 9 columnas distintas con 251 entradas en total. A partir de la función `info()` se puede observar que todas las columnas tienen algunos datos faltantes. En el otro extremo, la columna de `Apellido` no cuenta con ningún dato, por lo que una revisión de la causa sería precisa 

```python
# Eliminación de la columna sin datos
df.drop(columns=['Apellido'], inplace= True)
```

```python
# Conteo de NaN por columna
nan_counts = df.isna().sum()
print("Conteo de NaN por columna: ")
print(nan_counts)
```

```python
# Eliminación de filas con datos faltantes
df.dropna(inplace= True)
```

```python
df.info()
```

Actualmente todas las filas con algún dato faltante que es automáticamente reconocido por pandas como `NaN`han sido eliminadas del DataFrame, sin embargo, pueden seguir existiendo valores erróneos o algún tipo de dato faltante que no haya sido reconocido como tal. Por ende se realizará una revisión más exhaustiva sobre la naturaleza de los datos que quedan
