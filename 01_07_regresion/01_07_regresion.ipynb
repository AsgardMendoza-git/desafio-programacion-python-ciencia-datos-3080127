```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
regresion_lineal = LinearRegression()
```

```python
df = pd.read_csv('empleados.csv', delimiter=";")
```

```python
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df.isna()
```

```python
train = df.dropna()
```

```python
X = train.drop(columns=['Edad'])
y = train['Edad']
```

```python
regresion_lineal.fit(X,y)
regresion_lineal.score(X,y)
```

```python
regresion_lineal.coef_
```

```python
missing_values = df[df['Edad'].isna()]
```

```python
missing_values.shape
```

```python
X1 = missing_values.drop(columns=['Edad'])
y1 = regresion_lineal.predict(X1)
y1
```

```python
list(y1)
```

```python
missing_values.index
```

```python
imputation_policies = {}
j = 0 
for i in missing_values.index:

  imputation_policies[i] = int(y1[j])
  j += 1
```

```python
imputation_policies
```

```python
df['Edad'] = df['Edad'].fillna(imputation_policies)
```
